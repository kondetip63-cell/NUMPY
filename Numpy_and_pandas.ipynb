Create a 1D Numpy Array

```python
import numpy as np

temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])

print("Week 1 Temperatures:", temperatures_w1)
```

Inspection and properties

```python
print("Shape:", temperatures_w1.shape)
print("Data Type:", temperatures_w1.dtype)
print("Number of Elements:", temperatures_w1.size)
```

3. Array Operations

```python
temperatures_f = (temperatures_w1 * 9/5) + 32

print("Temperatures in Fahrenheit:", temperatures_f)
```

 maximum, minimum, and mean temperatures

```python
print("Maximum Temperature:", np.max(temperatures_w1))
print("Minimum Temperature:", np.min(temperatures_w1))
print("Mean Temperature:", np.mean(temperatures_w1))
```

4. Array Slicing and Indexing

```python
# First three days
first_three_days = temperatures_w1[:3]
print("First Three Days:", first_three_days)

# Weekend (last two days)
weekend = temperatures_w1[-2:]
print("Weekend Temperatures:", weekend)

# Middle three days (Days 3, 4, and 5)
middle_three_days = temperatures_w1[2:5]
print("Middle Three Days:", middle_three_days)
```

NumPy 2D Array

```python
import numpy as np

temperatures = np.array([
    [22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],
    [19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]
])

print("Temperatures Array:")
print(temperatures)
```

6. Inspect and Slice the 2D Array

```python
print("Shape:", temperatures.shape)
print("Data Type:", temperatures.dtype)
print("Total Elements:", temperatures.size)
```

B. Extract temperatures for each week and weekends

```python
# Week 1
print("Week 1:", temperatures[0, :])

# Week 2
print("Week 2:", temperatures[1, :])

# Weekend temperatures for both weeks
print("Weekend Temperatures:")
print(temperatures[:, -2:])
```

Part 2: Pandas Series

```python
import pandas as pd

marks = pd.Series(
    [95, 92, 89, 85, 80],
    index=['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5']
)

print(marks)
```

2. Indexing and Slicing

A. Access the 1st rank using integer position

```python
print(marks.iloc[0])
```

B. Retrieve the top 3 ranks using loc

```python
print(marks.loc[['Rank1', 'Rank2', 'Rank3']])
```

C. Retrieve the 3rd rank using iloc

```python
print(marks.iloc[2])
```

D. Filter ranks with marks greater than 90

```python
print(marks[marks > 90])
```

3. Manipulating the Series

A. Change the 1st rank's mark to 100

```python
marks.loc['Rank1'] = 100
print(marks)
```

B. Remove the last rank

```python
marks = marks.drop('Rank5')
print(marks)
```

C. Compute CGPA by dividing marks by 10

```python
cgpa = marks / 10
print(cgpa)
```

Part 3: Pandas DataFrame

```python
import pandas as pd

data = {
    'TransactionID': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'ProductCategory': [
        'Electronics', 'Clothing', 'Electronics', 'Furniture',
        'Clothing', 'Electronics', 'Furniture', 'Clothing',
        'Furniture', 'Electronics'
    ],
    'Region': [
        'North', 'South', 'North', 'East', 'West',
        'North', 'East', 'West', 'South', 'North'
    ],
    'Amount': [200, 150, 300, 450, 200, 250, 300, 180, 350, 400]
}

transactions = pd.DataFrame(data)

print(transactions)
```

2. Data Exploration

```python
# First 5 rows
print(transactions.head())

# Last 5 rows
print(transactions.tail())

# Shape
print("Shape:", transactions.shape)

# Column names
print("Columns:", transactions.columns)

# Data types and non-null counts
transactions.info()
```

B. Display only ProductCategory and Amount

```python
print(transactions[['ProductCategory', 'Amount']])
```

C. Retrieve the last 3 columns using iloc

```python
print(transactions.iloc[:, -3:])
```

D. Filter rows where Region is North and Amount is greater than 200

```python
result = transactions[
    (transactions['Region'] == 'North') &
    (transactions['Amount'] > 200)
]

print(result)
```

E. Count occurrences of each product category

```python
print(transactions['ProductCategory'].value_counts())
```

F. Find unique regions

```python
print(transactions['Region'].unique())
```

G. Find the mean amount for each region

```python
print(transactions.groupby('Region')['Amount'].mean())
```

3. Manipulating the DataFrame

```python
transactions.loc[
    transactions['TransactionID'] == 102, 'Amount'
] = 165

print(transactions)
```

B. Add a Discount column equal to 10% of Amount

```python
transactions['Discount'] = transactions['Amount'] * 0.10

print(transactions)
```

C. Remove the row with TransactionID 109

```python
transactions = transactions[
    transactions['TransactionID'] != 109
]

print(transactions)
```

D. Delete the Discount column

```python
transactions = transactions.drop(columns=['Discount'])

print(transactions)
```
